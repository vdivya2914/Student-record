```python
class Node:
    def __init__(self, roll):
        self.roll = roll
        self.left = None
        self.right = None

def insert(root, roll):
    if root is None:
        return Node(roll)
    if roll < root.roll:
        root.left = insert(root.left, roll)
    else:
        root.right = insert(root.right, roll)
    return root

def inorder(root):
    if root:
        inorder(root.left)
        print(root.roll, end=" ")
        inorder(root.right)

root = None

for roll in [50, 30, 70, 20, 40]:
    root = insert(root, roll)

print("Student Roll Numbers:")
inorder(root)
```
